```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2, 1000)

cna_cutoff = 4/16 * np.max(x)
a_cutoff = 1/7 * 4/16 * np.max(x)

x_flip = 2 - x   # reflection around x = 1

plt.plot(x_flip[x < (cna_cutoff - a_cutoff)], np.ones_like(x[x < (cna_cutoff - a_cutoff)]), color="blue")

mask = (x > (cna_cutoff - a_cutoff)) & (x < cna_cutoff)
plt.plot(x_flip[mask], np.ones_like(x[mask]), color="green")

plt.plot(x_flip[x > cna_cutoff], np.ones_like(x[x > cna_cutoff]), color="red")

plt.savefig("Figures/HP_approximation.png", bbox_inches="tight")
plt.show()
```
